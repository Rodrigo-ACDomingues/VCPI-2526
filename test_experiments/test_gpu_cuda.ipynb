```python
import torch

print("torch version:", torch.__version__)
print("cuda available:", torch.cuda.is_available())
print("cuda build:", torch.version.cuda)

if torch.cuda.is_available():
    x = torch.randn(2, 3, device="cuda")
    print("tensor device:", x.device)
```
